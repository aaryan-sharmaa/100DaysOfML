```python
from sklearn.datasets import make_regression
import numpy as np
import matplotlib.pyplot as plt

X, y = make_regression(
    n_samples=100,
    n_features=1,
    noise=10,
    random_state=42
)

X = X.ravel()
```

```python
w = 0
b = 0

learning_rate = 0.01
epochs = 50
```

```python
loss_history = []

for epoch in range(epochs):

    # Shuffle the data
    indices = np.random.permutation(len(X))
    X_shuffled = X[indices]
    y_shuffled = y[indices]

    for x_i, y_i in zip(X_shuffled, y_shuffled):

        # Prediction for ONE sample
        y_pred = w * x_i + b

        # Error
        error = y_i - y_pred

        # Gradients for ONE sample
        dw = -2 * x_i * error
        db = -2 * error

        # Update
        w = w - learning_rate * dw
        b = b - learning_rate * db

    # Calculate loss after the epoch
    y_pred_all = w * X + b
    loss = np.mean((y - y_pred_all) ** 2)
    loss_history.append(loss)
```

```python
from sklearn.datasets import make_regression
import matplotlib.pyplot as plt

X, y = make_regression(
    n_samples=100,
    n_features=1,
    noise=10,
    random_state=42
)

plt.scatter(X, y)
plt.xlabel("X")
plt.ylabel("y")
plt.show()
```

```python
X = X.ravel()
```

```python
w = 0
b = 0

learning_rate = 0.01
epochs = 1000

n = len(X)
loss_history = []

for i in range(epochs):

    # Prediction
    y_pred = w * X + b

    # Loss
    loss = np.mean((y - y_pred) ** 2)
    loss_history.append(loss)

    # Gradients
    dw = (-2 / n) * np.sum(X * (y - y_pred))
    db = (-2 / n) * np.sum(y - y_pred)

    # Update
    w = w - learning_rate * dw
    b = b - learning_rate * db

print("Weight:", w)
print("Bias:", b)
```

```python
y_pred = w * X + b

plt.scatter(X, y)
plt.plot(X, y_pred, color="red")
plt.show()
```

```python
plt.plot(loss_history)
plt.xlabel("Epoch")
plt.ylabel("MSE")
plt.show()
```

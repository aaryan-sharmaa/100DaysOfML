```python
import pandas as pd
import requests
from bs4 import BeautifulSoup
import numpy as np
```

```python
url = "https://www.ambitionbox.com/list-of-companies"

headers = {
    'User-Agent':'Mozilla/5.0'
}

webpage = requests.get(url, headers=headers).text
```

```python
# webpage=requests.get('https://www.ambitionbox.com/list-of-companies?page=1').text
```

```python
soup=BeautifulSoup(webpage,'lxml')
```

```python
# print(soup.prettify())
```

```python
company = soup.find_all('div', class_='companyCardWrapper')
```

```python
len(company)
```

```python
name = []
rating=[]
reviews=[]
salaries = []
interviews = []
jobs = []
benefits = []
photos = []
info=[]
highly_rated=[]
critical_rated=[]

for i in company:
    name.append(i.find('h2').text.strip())
    rating.append(i.find_all('div', class_="rating_text")[0].text.strip())
    reviews.append(i.find_all('span', class_="companyCardWrapper__companyRatingCount")[0].text.strip())
    actions = i.find_all('span', class_='companyCardWrapper__ActionCount')
    salaries.append(actions[1].text.strip())
    interviews.append(actions[2].text.strip())
    jobs.append(actions[3].text.strip())
    benefits.append(actions[4].text.strip())
    photos.append(actions[5].text.strip())
    info.append(
        i.find('span', class_='companyCardWrapper__interLinking').text.strip()
    )

#     ratings = i.find_all('span', class_='companyCardWrapper__ratingValues')
#     highly_rated.append(ratings[0].text.strip())
#     critical_rated.append(ratings[1].text.strip())
```

```python
df = pd.DataFrame({
    "Company": name,
    "Rating": rating,
    "Reviews": reviews,
    "Salaries": salaries,
    "Interviews": interviews,
    "Jobs": jobs,
    "Benefits": benefits,
    "Photos": photos,
    "Info": info
})
```
